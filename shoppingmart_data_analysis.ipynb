```python
master = master.drop_duplicates()
print(master.shape)
```

```python
master[master['quantity'].isnull()].index.intersection(master[master['sales_channel'].isnull()].index)
```

```python
master = master[master['quantity'] >= 0]
print(master.shape)
```

```python
master['sales_channel'] = master['sales_channel'].str.strip().str.lower()
master['sales_channel'] = master['sales_channel'].replace({
    'in store': 'in-store'
})
master['sales_channel'].unique()
```

```python
product['product_category'] = product['product_category'].fillna('Unknown')
product['product_price'] = product['product_price'].fillna(product['product_price'].median())
```

```python
product['product_category'] = product['product_category'].str.strip().str.lower()

product['product_category'] = product['product_category'].replace({
    'food & drinks': 'food and drinks',
    'health & wellness': 'health and wellness'
})

product['product_category'].unique()
```

```python
master['sales_channel'] = master['sales_channel'].str.strip().str.lower()
master['sales_channel'] = master['sales_channel'].replace({'in store': 'in-store'})
master['sales_channel'].unique()
```

```python
merged = master.merge(product, on='product_id', how='left')
print(merged.shape)
merged.isnull().sum()
```

```python
merged['product_category'] = merged['product_category'].fillna('unmatched')
merged['product_price'] = merged['product_price'].fillna('unmatched')
```

```python
merged[merged['product_category'] != 'unmatched']
```

```python
merged['sales_channel'] = merged['sales_channel'].fillna('unknown')
```

**KPI**

```python
total_revenue = merged['total_bill'].sum()
print("Total Revenue:", total_revenue)

revenue_by_category = merged.groupby('product_category')['total_bill'].sum().sort_values(ascending=False)
print(revenue_by_category)

revenue_by_channel = merged.groupby('sales_channel')['total_bill'].sum()
print(revenue_by_channel)

aov = merged['total_bill'].sum() / merged['order_id'].nunique()
print("Average Order Value:", aov)

data_quality_rate = (merged['product_category'] != 'unmatched').mean() * 100
print("Data Quality Rate (%):", data_quality_rate)
```
